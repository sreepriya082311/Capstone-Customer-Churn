```python
# Replace missing TotalCharges with 0
df["TotalCharges"] = df["TotalCharges"].fillna(0)

# Verify missing values are handled
print("Missing TotalCharges:", df["TotalCharges"].isnull().sum())

# Check data type
print("TotalCharges data type:", df["TotalCharges"].dtype)
```

```python
# Convert TotalCharges to numeric format
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"])

# Verify the data type
print("TotalCharges data type:", df["TotalCharges"].dtype)

# Check missing values again
print("\nMissing Values:")
print(df.isnull().sum().sum())

# Check dataset shape
print("\nFinal Dataset Shape:", df.shape)
```

```python
# Save the cleaned dataset
df.to_csv("../data/cleaned_customer_churn.csv", index=False)

print("Cleaned dataset saved successfully!")
```

```python
# Business Question 1: Customer Count and Churn

total_customers = len(df)

churned_customers = (df["Churn"] == "Yes").sum()

not_churned_customers = (df["Churn"] == "No").sum()

print("Total Customers:", total_customers)
print("Churned Customers:", churned_customers)
print("Customers Retained:", not_churned_customers)
```

```python
# Business Question 2: Churn and Retention Rate

churn_rate = (churned_customers / total_customers) * 100
retention_rate = (not_churned_customers / total_customers) * 100

print("Customer Churn Rate:", round(churn_rate, 2), "%")
print("Customer Retention Rate:", round(retention_rate, 2), "%")
```

```python
# Business Question 3: Churn by Contract Type

contract_churn = df.groupby("Contract")["Churn"].value_counts().unstack(fill_value=0)

print("Customer Churn by Contract Type:")
print(contract_churn)
```

```python
# Business Question 4: Average Monthly Charges

avg_charges = df.groupby("Churn")["MonthlyCharges"].mean()

print("Average Monthly Charges by Churn Status:")
print(avg_charges.round(2))
```

```python
# Business Question 5: Average Tenure by Churn Status

avg_tenure = df.groupby("Churn")["tenure"].mean()

print("Average Customer Tenure by Churn Status:")
print(avg_tenure.round(2))
```

```python
# Business Question 6: Churn by Internet Service

internet_churn = df.groupby("InternetService")["Churn"].value_counts().unstack(fill_value=0)

print("Churn by Internet Service:")
print(internet_churn)
```

```python
# Create a derived column for customer tenure groups

df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=[-1, 12, 24, 48, 72],
    labels=["0-12 Months", "13-24 Months", "25-48 Months", "49-72 Months"]
)

# Check the new column
print(df[["tenure", "TenureGroup"]].head(10))

print("\nTenure Group Counts:")
print(df["TenureGroup"].value_counts().sort_index())
```

```python
# Check outliers using the IQR method

columns_to_check = ["MonthlyCharges", "TotalCharges"]

for col in columns_to_check:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower_limit) | (df[col] > upper_limit)).sum()

    print(f"{col}")
    print("Lower Limit:", round(lower_limit, 2))
    print("Upper Limit:", round(upper_limit, 2))
    print("Outliers Count:", outliers)
    print()
```

```python
# Numerical Exploratory Data Analysis (EDA)

numeric_columns = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]

print("Numerical Statistics:")
print(df[numeric_columns].describe().round(2))

print("\nAverage Monthly Charges:", round(df["MonthlyCharges"].mean(), 2))
print("Median Monthly Charges:", round(df["MonthlyCharges"].median(), 2))
print("Average Total Charges:", round(df["TotalCharges"].mean(), 2))
print("Average Tenure:", round(df["tenure"].mean(), 2))
```
